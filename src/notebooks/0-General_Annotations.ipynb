# __On Computation Complexity__
The normal equation computes the inverse of $X^TX$, which is an (n + 1) x (n + 1) matrix (where n is the number of features). The computational complexity of inverting those matrices is typically about $O(n^2.4)$ to $O(n^3)$.

The `SVD` from SKLearn is about $O(n^2)$.

Both the normal equation and SVD get slow when the number of __features__ grows large. On the other side, both are linear regarding to the number of instances in the training set ($O(n)$), so they handle large traning sets efficiently.

When working with large number of features, a good approach is to use the famous __Gradient Descent__.

### __Gradient Descent__
It is a generic optimization algorithm that finds optimal solutions. The general idea is to tweak parameters iteratively in order to minimize a cost function.

The `gradient descent` measures the local gradient of the error function with regard to the parameter vector __$\theta$__, and it goes in the direction of descending gradient. Once the gradient is zero, we have reached a minimum.

We start by filling __$\theta$__ with random values (_random initialization_). Then we improve those values gradually, taking one little step at a time, each step attempting to decrease the cost function, until the algorithm converges to a minimum.

An important parameter in the gradient is the __Learning Rate__ hyperparameter. It is the size of each step. If we set the learning rate to a very small number, the algorithm will have to go through many iterations to converge, which can take a long time. On the other hand, if we set it too high, the gradiant will probably jump and end up on the other side. This can make the algorithm divergem failing to find a good solution.

When using gradient descent, we should ensure that all features have a similar scale.

### __Batch Gradient Descent__
Most models have more than one parameter. Therefore, to implement gradient descent, we need to compute the gradient of the cost function with regard to each model parameter $\theta j$. We need to calculate how much the cost function will change if we change $\theta j$ a little bit. This is _partial derivative_.

But instead of computing partial derivatives individually, we can compute them all in one go with the __Gradient Vector__. The gradient vector contains all the partial derivatives of the cost function.

This computation involves calculations over the full training set $X$, at each gradient step. This is why it is called _batch gradient descent_. It uses the whole batch of training data at every step. Because of that, it is very slow on very large training sets.
